```python
X_train = train.iloc[:,1:40].values
Y_train = train.iloc[:,-1].values
# X_test = test.iloc[:,3:15].values
# Y_test = test.iloc[:,-1].values
```

```python
train=train.replace({'cold': 1, 'noncold': 0})
#test=test.replace({'cold': 1, 'noncold': 0})
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
                        X_train, Y_train, test_size = 0.20)
```

```python
from sklearn.preprocessing import  MinMaxScaler #MinMaxScaler #StandardScaler
scaler = MinMaxScaler()
scaler.fit(X_train)
X_train =scaler.transform(X_train)
X_test =scaler.transform(X_test)
```

```python
model = SVC(kernel= 'rbf', gamma='auto', C=200, class_weight= 'balanced')
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
print('Accuracy: %.3f' % accuracy_score(Y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
matrix = classification_report(Y_test, y_pred)
print('Classification report : \n',matrix)
```
